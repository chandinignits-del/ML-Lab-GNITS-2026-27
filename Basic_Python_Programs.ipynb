**A .Write a Python Program to Perform Arithmetic Operations on different types given through input method in python**

```python
while(True):
  n=int(input("Enter choice: "))
  num1=int(input("Enter num1: "))
  num2=int(input("Enter num1: "))
  if(n==1):
    print(num1+num2)
  elif(n==2):
    print(num1-num2)
  elif(n==3):
    print(num1//num2)
  elif(n==4):
    print(num1/num2)
  elif(n==5):
    print(num1*num2)
  else:
    break
```

**B.Write a program to check whether a given string is palindrome or not.**

```python
n=input()
ls=list(n)
rs=ls.reverse()
ss=str(rs)
if(n==ss):
  print("Palindrome")
else:
  print("Not a Palindrome")
```

**E.Write a python to count the occurrences of word in a text file using dictionaries**

```python
def count_word_occurrences(file_path):
  try:
    with open(file_path, 'r') as file:
      text = file.read().lower()
      words = text.split()
      word_count = {}
      for word in words:
        word = word.strip(",.!?\"'()[]{}")
        word_count[word] = word_count.get(word, 0) + 1
      return word_count
  except FileNotFoundError:
    print("The file was not found. Please check the file path.")
    return {}
file_path = "example.txt"
word_occurrences = count_word_occurrences(file_path)
if word_occurrences:
  for word, count in word_occurrences.items():
    print(f"'{word}': {count}")
```

**F.Write a program to perform different basic operations on a tuple.**

```python
my_tuple = (10, 20, 30, 40, 50)
print("First element:", my_tuple[0])
print("Last element:", my_tuple[-1])
print("Slice (2nd to 4th):", my_tuple[1:4])
print("Length of tuple:", len(my_tuple))
print("Is 20 in tuple?", 20 in my_tuple)
new_tuple = my_tuple + (60, 70)
print("After concatenation:", new_tuple)
repeated_tuple = my_tuple * 2
print("After repetition:", repeated_tuple)
print("Iterating through tuple:")
for item in my_tuple:
  print(item)
print("Index of 30:", my_tuple.index(30))
print("Count of 20:", my_tuple.count(20))
```
