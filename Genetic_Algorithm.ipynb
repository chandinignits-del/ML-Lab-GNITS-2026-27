```python
import random
# Parameters
BIT_LENGTH = 5
POP_SIZE = 20
GENERATIONS = 10
MUTATION_RATE = 0.1
def bin_to_int(binary_str):
  return int(binary_str, 2)
def fitness(individual):
  x = bin_to_int(individual)
  return x * x
def random_individual():
  return ''.join(random.choice('01') for _ in range(BIT_LENGTH))
def select(population, fitnesses):
  total = sum(fitnesses)
  pick = random.uniform(0, total)
  current = 0
  for ind, fit in zip(population, fitnesses):
    current += fit
    if current > pick:
      return ind
  return population[-1]
def crossover(p1, p2):
  point = random.randint(1, BIT_LENGTH - 1)
  return p1[:point] + p2[point:], p2[:point] + p1[point:]
def mutate(individual):
  mutated = ''
  for bit in individual:
    if random.random() < MUTATION_RATE:
      mutated += '1' if bit == '0' else '0'
    else:
      mutated += bit
  return mutated
def genetic_algorithm():
  population = [random_individual() for _ in range(POP_SIZE)]
  best_solution = None
  best_fitness = -1
  for gen in range(GENERATIONS):
    fitnesses = [fitness(ind) for ind in population]
    max_fit = max(fitnesses)
    max_ind = population[fitnesses.index(max_fit)]
    if max_fit > best_fitness:
      best_fitness = max_fit
      best_solution = max_ind
      print(f"GA Gen {gen+1}: Best fitness = {max_fit} ({bin_to_int(max_ind)})")
      new_population = []
      while len(new_population) < POP_SIZE:
       p1 = select(population, fitnesses)
       p2 = select(population, fitnesses)
       c1, c2 = crossover(p1, p2)
       new_population.append(mutate(c1))
       if len(new_population) < POP_SIZE:
         new_population.append(mutate(c2))
    population = new_population
  return best_solution, best_fitness
def random_search():
  best_ind = None
  best_fit = -1
  evaluations = POP_SIZE * GENERATIONS
  for _ in range(evaluations):
    ind = random_individual()
    fit = fitness(ind)
    if fit > best_fit:
      best_fit = fit
      best_ind = ind
  print(f"Random Search: Best fitness = {best_fit} ({bin_to_int(best_ind)})")
  return best_ind, best_fit
# Run and compare
print("Running Genetic Algorithm:")
ga_sol, ga_fit = genetic_algorithm()
print("\nRunning Random Search:")
rs_sol, rs_fit = random_search()
print(f"\nGA best solution: {ga_sol} ({bin_to_int(ga_sol)}) fitness {ga_fit}")
print(f"Random Search best solution: {rs_sol} ({bin_to_int(rs_sol)}) fitness {rs_fit}")
```
