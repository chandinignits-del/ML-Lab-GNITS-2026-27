Week 2: Implement FIND-S algorithm for finding the most specific hypothesis based on a given set of training data samples.

```python
import csv
hypo = ['%','%','%','%','%','%'];

with open('trainingdata.csv') as csv_file:
    readcsv = csv.reader(csv_file, delimiter=',')
    print(readcsv)

    data = []
    print("\nThe given training examples are:")
    for row in readcsv:
        print(row)
        if row[len(row)-1].upper() == "YES":
            data.append(row)
```

```python
print("\nThe positive examples are:");
for x in data:
    print(x);
print("\n");
```
